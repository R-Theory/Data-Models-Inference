## **Topic 2: Data Collection & Sampling Strategies (Sections 1.3 and 1.4)**


### *Key Terminology*
Population = The ENTIRE group you want to study/make conclusions about

Examples: All UNC students, all swordfish in the Atlantic, all US counties

Sample = A SUBSET of the population that you actually collect data from

We use samples to make "educated guesses" about populations

Census = Collecting data from the ENTIRE population (rare and expensive!)
Survey = Collecting data from a sample
Bias = Systematic error that makes your sample unrepresentative of the population

### Good Sampling Techniques (Reduce Bias)

| Sampling Technique | Method | Example | When to Use | Notes |
|-------------------|--------|---------|-------------|-------|
| **Simple Random Sample (SRS)** | Every member of the population has an EQUAL chance of being selected | Write all names on slips of paper, put in hat, draw randomly | Gold standard! Use when you can access the whole population easily | Downside: Can be expensive if population is spread out geographically |
| **Stratified Sampling** | 1. Divide population into strata (groups with similar characteristics)<br>2. Take a Simple Random Sample from EACH stratum<br>Common strata: Age groups, gender, income level, geographic region | Survey UNC students by randomly sampling from each year (freshman, sophomore, junior, senior) | When you have a characteristic you think will substantially influence results | Goal: Ensure you don't accidentally omit or underrepresent key subgroups |
| **Cluster Sampling** | 1. Divide population into clusters (geographic or natural groups)<br>2. Randomly select a FEW clusters<br>3. Sample EVERYONE in those selected clusters | Study malaria in 30 villages → randomly pick 5 villages, test everyone in those 5 | Large, geographically spread populations where clusters are diverse WITHIN themselves | Downside: Need advanced statistical methods to analyze |
| **Multistage Sampling** | Like cluster sampling, but take a random sample WITHIN each selected cluster (not everyone) | Pick 5 villages, then randomly sample 20 people from each village | When clusters are too big to sample everyone | Combines cluster and random sampling approaches |


### Bad Sampling Techniques
| Sampling Technique | Description | Example | Problem |
|-------------------|-------------|---------|---------|
| **Convenience Sample** | Sampling whoever is easily accessible | Surveying only people walking by you in the Pit | Not representative! Misses people who aren't in that location |
| **Voluntary Response** | Only people who CHOOSE to respond are included | Online reviews, call-in polls | People with strong opinions are more likely to respond |
| **Non-Response Bias** | When many people in your sample don't respond | Send survey to 1000 people, only 200 respond | The 200 who responded might be different from the 800 who didn't |

## **Topic 3: Experiments**


### Four Principles of Experimental Design
1. CONTROL
- Assign participants to MULTIPLE treatment groups
- At least one group should be control (no treatment)
- Purpose: Account for confounding variables
- Treatment = explanatory variable you're manipulating
2. RANDOMIZE
- Assign subjects RANDOMLY to treatment groups
- Purpose: Reduce bias, balance out unknown confounding variables
- How: Coin flip, random number generator, etc.
#### Special Case: Matched Pairs
- Use "similar" subjects, put one in control, one in treatment.
- BEST matched pairs: Identical twins!
- Still randomly assign which twin gets which treatment
3. REPLICATE
- Test on MULTIPLE subjects
- More subjects = better!
- Must be repeatable by other scientists
- If not repeatable → not believable!
4. BLOCK
- "Experimental equivalent of stratifying"
- Blocking variable: Characteristic likely to influence outcomes
    - NOT an explanatory variable
    - NOT what you're testing
- Break subjects into blocks BEFORE random assignment
- Then randomly assign within each block

## Topic 4: Numerical Data Analysis

### Visualizing Distributions: Histograms
Histogram: Shows distribution of ONE numerical variable

Key Features:
- Bins (classes/ranges) on x-axis
  - Equal width, no gaps, no overlap
  - Must cover all data
  - Choice of bin width matters!
- Frequency or Relative Frequency on y-axis
  - Frequency = count
  - Relative Frequency = count/n (always sums to 1)
- Bars are adjacent (no gaps)
  - Unlike bar plots for categorical data

> Shape DOES NOT change when using frequency vs relative frequency!

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt

X = np.arange(0,1,0.02)
np.random()

plt.figure(figsize=(10,7))
plt.hist(x=X)
```
